Channel Manipulation

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img_array = np.random.randint(0, 255, (300, 300, 3), dtype='uint8')
user_inp = int(input("Which color you wnt to remove? (For Red= 0, Green= 1, and Blue= 2): "))
```

```python
if user_inp == 0:
    no_red = img_array.copy()
    no_red[:,:, 0] = 0
    plt.imshow(no_red)
    plt.title("Red Color Removed")
    plt.show()
    print(f"Shape: {no_red.shape}")
elif user_inp == 1:
    no_green = img_array.copy()
    no_green[:, :, 1] = 0
    plt.imshow(no_green)
    plt.title("Green Color Removed")
    plt.show()
    print(f"Shape: {no_green.shape}")
elif user_inp == 2:
    no_blue = img_array.copy()
    no_blue[:, :, 2] = 0
    plt.imshow(no_blue)
    plt.title("Blue Color Removed")
    plt.show()
    print(f"Shape: {no_blue.shape}")
else:
    print("Invalide index!")
```
