```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image  # Image load karne ke liye

# 1. PC se Image Load karein
# 'test_image.jpg' ki jagah apni kisi bhi image ka naam likhein
try:
    img = Image.open('Vango Art.jpg') 
    img_array = np.array(img) # Image ko Numpy array mein badal diya
except:
    print("Error: Image file nahi mili! File ka naam sahi check karein.")
    # Agar file nahi hai to hum practice ke liye random colors bana lete hain
    img_array = np.random.randint(0, 255, (300, 300, 3), dtype='uint8')

# 2. Image ka structure samjhein
print(f"Shape: {img_array.shape}") # (Height, Width, Channels) - Channels 3 hote hain (R, G, B)

# 3. Grayscale (Black & White) Conversion Logic
# B&W karne ka formula: (Red + Green + Blue) / 3
# Hum axis=2 use karte hain kyunki RGB channels 3rd dimension mein hain
grayscale = np.mean(img_array, axis=2)

# 4. Color Filters (Red Filter)
# Hum Green aur Blue channels ko zero kar denge
red_filter = img_array.copy()
red_filter[:, :, 1] = 0 # Green channel = 0
red_filter[:, :, 2] = 0 # Blue channel = 0

# 5. Visualizing with Matplotlib
plt.figure(figsize=(15, 5)) # Window ka size set kiya

# Original Image
plt.subplot(1, 3, 1) # 1 row, 3 columns, position 1
plt.imshow(img_array)
plt.title("Original RGB")
plt.axis('off')

# Grayscale Image
plt.subplot(1, 3, 2) # position 2
plt.imshow(grayscale, cmap='gray')
plt.title("Black & White (Mean)")
plt.axis('off')

# Red Filtered Image
plt.subplot(1, 3, 3) # position 3
plt.imshow(red_filter)
plt.title("Only Red Channel")
plt.axis('off')

plt.show()
```
